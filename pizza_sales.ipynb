```python
#Find Total Sales by Hour--(Total Pizza Quantity Sold in Each Hour)

#Find Busiest Hour---(Hour with Maximum Pizza Sales)
busiest_hour=df.groupby('hour')['quantity'].sum().sort_values(ascending=False).head(1)
print("Busiest Hour in pizza sales is",busiest_hour)

#Find Least Busy Hour---(Hour with Minimum Pizza Sales)
Least_busy_hour=df.groupby('hour')['quantity'].sum().sort_values().head(1)
print(Least_busy_hour)


#Find Revenue by Hour
hour_revenue=df.groupby('hour')['total_price'].sum().sort_values(ascending=False)
print(hour_revenue)

#Total Pizza Sold by Month
month_pizza=df.groupby('month')['quantity'].sum()
print(month_pizza)

#Total Revenue by Month
month_revenue=df.groupby('month')['total_price'].sum()
print(month_revenue)

#Find Month with Highest Sales
month_pizza=df.groupby('month_name')['quantity'].sum().sort_values(ascending=False).head(1)
print(month_pizza)

#Find Month with Lowest Sales
month_pizza=df.groupby('month_name')['quantity'].sum().sort_values().head(1)
print(month_pizza)

#Find Month with Highest Revenue
month_highest_revenue=df.groupby('month_name')['total_price'].sum().sort_values(ascending=False).head(1)
print(month_highest_revenue)

#Find Month with Lowest Revenue
month_Lowest_revenue=df.groupby('month_name')['total_price'].sum().sort_values().head(1)
print(month_Lowest_revenue)

#Total Pizza Sold by Day of Week
week_pizza=df.groupby('day_name')['quantity'].sum()
print(week_pizza)

#Total Revenue by Day of Week
week_pizza=df.groupby('day_name')['total_price'].sum()
print(week_pizza)

#Find Busiest Day
busiest_day=df.groupby('day')['order_id'].nunique().sort_values(ascending=False).head(1)
print("busiest_day is",busiest_day)

#Find Least Busy Day
Least_busiest_day=df.groupby('day')['order_id'].nunique().sort_values().head(1)
print(Least_busiest_day)

#Find Highest Revenue Day
Revenue_day=df.groupby('day')['total_price'].sum().sort_values(ascending=False).head(1)
print(Revenue_day)

#Find Lowest Revenue Day
Revenue_day=df.groupby('day')['total_price'].sum().sort_values().head(1)
print(Revenue_day)

#Total Orders by Month
order_month=df.groupby('month')['order_id'].nunique()
print(order_month)

#Total Orders by Day of Week
order_week=df.groupby('day_name')['order_id'].nunique()
print(order_week)

#Top 5 Dates with Highest Revenue
Revenue_date=df.groupby('order_date')['total_price'].sum().sort_values(ascending=False).head(5)
print(Revenue_date)

#Top 5 Dates with Highest Sales
Revenue_date=df.groupby('order_date')['quantity'].sum().sort_values(ascending=False).head(5)
print(Revenue_date)
```
